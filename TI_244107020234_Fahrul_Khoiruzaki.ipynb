**Keguanaan Masing Masing Library:**

a. PyPREP

PyPREP digunakan untuk membantu mempersiapkan dan membersihkan data sebelum dilakukan analisis. Library ini dapat membantu melihat kondisi data, seperti data yang kosong, data yang tidak sesuai, atau masalah lain pada dataset. Jadi, data yang akan digunakan untuk proses selanjutnya bisa lebih rapi dan siap diolah.

b. Scipy

SciPy merupakan library Python yang digunakan untuk perhitungan ilmiah dan matematika. Library ini menyediakan berbagai fungsi untuk statistik, optimasi, aljabar linear, integrasi, dan pengolahan sinyal. SciPy biasanya digunakan ketika perhitungan yang dibutuhkan lebih kompleks dibandingkan perhitungan dasar Python.

c. wandb

wandb digunakan untuk mencatat dan memantau proses eksperimen machine learning. Misalnya, kita bisa menyimpan nilai akurasi, loss, parameter model, serta hasil dari beberapa percobaan. Dengan begitu, kita lebih mudah membandingkan eksperimen dan mengetahui model mana yang memberikan hasil terbaik.

d. PyECG

pyECG digunakan untuk mengolah dan menganalisis sinyal ECG (Electrocardiogram) atau sinyal elektrokardiogram dari jantung. Library ini dapat membantu dalam proses pengolahan sinyal ECG sehingga informasi mengenai aktivitas listrik jantung dapat dianalisis lebih lanjut.
